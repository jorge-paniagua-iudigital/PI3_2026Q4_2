```python
import pandas as pd
import numpy as np
from pathlib import Path
import warnings

warnings.filterwarnings("ignore")

# ---------------------------------------------------------
# 1. VALIDACIÓN ANTES DE LA TRANSFORMACIÓN
# ---------------------------------------------------------
print("=== 1. ESTADO ANTES DE TRANSFORMACIÓN ===")
print(f"Filas iniciales: {len(df)}")
print(f"Columnas iniciales ({len(df.columns)}): {list(df.columns)}")
print(f"Valores nulos totales: {df.isnull().sum().sum()}")
print(f"Duplicados exactos: {df.duplicated().sum()}\n")

# ---------------------------------------------------------
# 2. ENRIQUECIMIENTO DE VARIABLES
# ---------------------------------------------------------
df_enriquecido = df.copy()

# A. Extraer Año y Trimestre a partir de Period (ej: '2015 Q1')
df_enriquecido["Año"] = df_enriquecido["Period"].str.split(" ").str[0].astype(int)
df_enriquecido["Trimestre"] = df_enriquecido["Period"].str.split(" ").str[1]

# B. Crear variable escalada en Miles de Millones (Billion USD/COP)
df_enriquecido["Valor_Miles_Millones"] = df_enriquecido["Value - Randomized"] / 1e9

# C. Calcular participación porcentual de cada subcategoría por periodo y categoría
totales_periodo = df_enriquecido.groupby(["Period", "Category"])["Value - Randomized"].transform("sum")
df_enriquecido["Participacion_Categoria_Pct"] = np.where(
    totales_periodo > 0,
    (df_enriquecido["Value - Randomized"] / totales_periodo) * 100,
    0.0
)

# ---------------------------------------------------------
# 3. VALIDACIÓN DESPUÉS DE LA TRANSFORMACIÓN
# ---------------------------------------------------------
print("=== 2. ESTADO DESPUÉS DE ENRIQUECIMIENTO ===")
print(f"Filas finales: {len(df_enriquecido)}")
print(f"Columnas finales ({len(df_enriquecido.columns)}): {list(df_enriquecido.columns)}")
print(f"Nuevas variables incorporadas: ['Año', 'Trimestre', 'Valor_Miles_Millones', 'Participacion_Categoria_Pct']\n")

# ---------------------------------------------------------
# 4. TABLAS DE RESULTADOS PARA RESPONDER PREGUNTAS DE NEGOCIO
# ---------------------------------------------------------

# Pregunta 1: ¿Cuáles son las subcategorías con mayor aporte financiero promedio y acumulado?
tabla_subcat = df_enriquecido.groupby(["Category", "Sub Cat"]).agg(
    Ingreso_Total_MM=("Valor_Miles_Millones", "sum"),
    Ingreso_Promedio_Trimestral_MM=("Valor_Miles_Millones", "mean"),
    Ingreso_Mediano_MM=("Valor_Miles_Millones", "median"),
    Participacion_Promedio_Pct=("Participacion_Categoria_Pct", "mean")
).reset_index().sort_values(by=["Category", "Ingreso_Total_MM"], ascending=[True, False])

# Pregunta 2: ¿Cómo han evolucionado los ingresos anuales entre regiones y productos?
tabla_anual = df_enriquecido.groupby(["Año", "Category"]).agg(
    Ingreso_Anual_MM=("Valor_Miles_Millones", "sum")
).reset_index().pivot(index="Año", columns="Category", values="Ingreso_Anual_MM").reset_index()

# ---------------------------------------------------------
# 5. EXPORTACIONES AUTOMÁTICAS (.CSV y .TXT)
# ---------------------------------------------------------
ruta_datos = Path("datos")
ruta_reportes = Path("reportes")
ruta_datos.mkdir(parents=True, exist_ok=True)
ruta_reportes.mkdir(parents=True, exist_ok=True)

# Exportar dataset preparado
df_enriquecido.to_csv("datos/company_dataset_preparado.csv", index=False, encoding="utf-8")

# Exportar tablas de resultados
tabla_subcat.to_csv("reportes/tabla_desempeno_subcategorias.csv", index=False, encoding="utf-8")
tabla_anual.to_csv("reportes/tabla_evolucion_anual.csv", index=False, encoding="utf-8")

# Regenerar reporte descriptivo .txt actualizado
ruta_txt = reportes.guardar_reporte(df_enriquecido, "reportes/company_dataset_summary_ae3.txt")

print(f" Archivos exportados automáticamente:")
print(f" - Dataset preparado: datos/company_dataset_preparado.csv")
print(f" - Tabla Subcategorías: reportes/tabla_desempeno_subcategorias.csv")
print(f" - Tabla Evolución Anual: reportes/tabla_evolucion_anual.csv")
print(f" - Resumen .txt: {ruta_txt}\n")

print("--- VISTA PREVIA TABLA SUBCATEGORÍAS ---")
print(tabla_subcat.round(2).to_string(index=False))

print("\n--- VISTA PREVIA EVOLUCIÓN ANUAL ---")
print(tabla_anual.round(2).to_string(index=False))
```
